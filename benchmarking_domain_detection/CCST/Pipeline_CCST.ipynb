```python
def save_file(data_path, results_path, data_name):

    adata = sc.read_h5ad(data_path)
   
    tmp = pd.read_table("./results/" + data_name + "/lambdaI0.3/types.txt" , header = None)
    
    z = pd.DataFrame([adata.obs.index,tmp[1]]).T
    z = z.rename(columns = {0:'Id', 1:'Expected'})
    z.set_index('Id', drop = True).to_csv(results_path)
    
    img = mpimg.imread("./results/" + data_name + "/lambdaI0.3/spacial.png")
    plt.imshow(img)
    plt.show()
```
