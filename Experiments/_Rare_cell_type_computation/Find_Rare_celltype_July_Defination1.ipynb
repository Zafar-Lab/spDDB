### Criteria 1 : Bottom 10% in terms of CTP.

```python
def get_rare(ctp_path):
    dt = pd.read_csv(ctp_path)
    celltypes = dt.columns[1:].values
    print (celltypes, len(celltypes), "\n")
    dt = dt[celltypes]
    dt = dt.mean()
    rare = dt.sort_values()[: round(0.1*len(dt))]
    return rare
```
