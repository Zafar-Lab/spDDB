```python

bi_metrics = ['morans_r', 'spatial_pearson', 'lee_stat','ssim', 'geary_c']

acc_metrics = ['cosine_sim', 'pearson']

methods_all = ['Autogenes','CARD','Cell2location','CellDART','CelloScope','DestVI','GraphST','Polaris','RCTD',
               'Redeconve','SD','Seurat','SONAR','SpatialDecon','SpatialDWLS','SpiceMix','Spotlight','STDeconvolve',
               'Stereoscope','STRIDE','Tangram']

spot_metrics = ["RMSE", "JSD"]

cc_path = "/home/ajita/Downloads/benchmarking_excels/Cell_Charter_metrics.xlsx"
sheet_path = "/home/ajita/Downloads/benchmarking_excels/Spatial-Deconvolution Methods.xlsx"
```

```python
"""
# Save all composite score files in Dataset \Metrics
# Save funkyheatmap plots in Dataset \Plots
"""

def compute(path, slice_id, dataset_name):

    if len(slice_id) > 1:
        metrics_path =  path + slice_id + "/Metrics/"
    else:
        metrics_path = path + "Metrics/"
        
    data_csv_path = metrics_path + "data.csv"
    col_group_csv_path = metrics_path + "column_groups.csv"
    col_info_csv_path = metrics_path + "columns.csv"
    row_group_csv_path = metrics_path + "row_groups.csv"
    row_info_csv_path = metrics_path + "rows.csv"
    
    create_composite_score_file(metrics_path, cc_path, sheet_path, dataset_name, methods_all, bi_metrics, acc_metrics, spot_metrics,
               data_csv_path)
    
    create_plotting_files(methods_all, col_group_csv_path, col_info_csv_path, row_group_csv_path, row_info_csv_path)
```
