### DLPFC

```python
%%R

path <- "/data/Ajita/Spatial/Datasets/Spatial_Deconvolution/DLPFC/ST/"
sim_path <- "/data/Ajita/Spatial/Datasets/Spatial_Deconvolution/DLPFC/simulator_outputs/simulated_gene_expression"

# 151508, 151509, 669 -> kernel died, 
#"151670", "151673", "151674"
for (slice_id in list("151508", "151509", "151670", "151673", "151674")) {
  ddsList <- list()
  orig_data_path <- paste0(paste0(path, slice_id), "/ST.h5ad")
  sim_data_path <- paste0(paste0(sim_path, slice_id), "_average.h5ad")
  
  dds1 <- convert_ann2dds_original(orig_data_path, sim_data_path)
  dds2 <- convert_ann2dds(sim_data_path)
  name1 <- paste0("Orig_", slice_id)
  name2 <- paste0("Sim_", slice_id)
  
  ddsList[[ paste0("Orig_", slice_id)]] <- dds1
  ddsList[[ paste0("Sim_", slice_id)]] <- dds2
  #ddsList <- c( ddsList, list(name1 = dds1, name2 = dds2))
  countsim_plots(ddsList, paste0("DLPFC_", slice_id))
  individual_plots(paste0("DLPFC_", slice_id))
}
```
