```python
def get_all_dataset_paths(dataset_names):
    """
    Given a list of dataset names, return a dictionary mapping each dataset
    name to its paths dictionary (st_path, gnd_path if applicable).
    
    Parameters
    ----------
    dataset_names : list of str
        Dataset names.
    
    Returns
    -------
    dict
        {
            dataset_name1: {"st_path": ..., "gnd_path": ...},
            dataset_name2: {"st_path": ...},
            ...
        }
    """
    
    def _single_dataset_paths(dataset_name):
        if dataset_name.startswith("DLPFC"):
            dataset_id = dataset_name.replace("DLPFC", "")
            return {
                "st_path": f"/data/Ajita/Spatial/Datasets/Spatial_Domain/DLPFC/{dataset_id}/ST.h5ad",
                "gnd_path": f"/data/Ajita/Spatial/Datasets/Spatial_Domain/DLPFC/{dataset_id}/metadata.tsv"
            }

        elif dataset_name.startswith("embryo"):
            dataset_id = dataset_name.replace("embryo", "")
            return {
                "st_path": f"/data/Ajita/Spatial/Datasets/Spatial_Domain/embryo/E{dataset_id}_E1S1.MOSTA.h5ad"
            }

        elif dataset_name == "mouse_breast_cancer":
            return {
                "st_path": "/data/Ajita/Spatial/Datasets/Spatial_Domain/mouse_breast_cancer/mouse_breast_cancer.h5ad"
            }

        elif dataset_name.startswith("MERFISH_brain"):
            dataset_id = dataset_name.replace("MERFISH_brain", "")
            return {
                "st_path": f"/data/Ajita/Spatial/Datasets/Spatial_Domain/MERFISH_brain/MERFISH_{dataset_id}.h5ad"
            }

        elif dataset_name == "simulated_breast_atlas":
            return {
                "st_path": "/data/Ajita/Spatial/Datasets/Spatial_Domain/Simulated_Breast_Atlas_ST8/simulated_gene_expression_average.h5ad",
                "gnd_path": "/data/Ajita/Spatial/Datasets/Spatial_Domain/Simulated_Breast_Atlas_ST8/simulated_ground_truth_ST_8.csv"
            }

        elif dataset_name == "osmFISH":
            return {
                "st_path": "/data/Ajita/Spatial/Datasets/Spatial_Domain/osmFISH/osmfish.h5ad"
            }

        elif dataset_name.startswith("simulated_kidney_cancer"):
            dataset_id = dataset_name.replace("simulated_kidney_cancer", "")
            return {
                "st_path": f"/data/Ajita/Spatial/Datasets/Spatial_Domain/Simulated_kidney_cancer/simulated_gene_expression_{dataset_id}_average.h5ad",
                "gnd_path": f"/data/Ajita/Spatial/Datasets/Spatial_Domain/Simulated_kidney_cancer/simulated_ground_truth_{dataset_id}.csv"
            }

        elif dataset_name.startswith("simulated_breast_cancer"):
            dataset_id = dataset_name.replace("simulated_breast_cancer", "")
            return {
                "st_path": f"/data/Ajita/Spatial/Datasets/Spatial_Domain/Simulated_Breast_Cancer/{dataset_id}/simulated_gene_expression_average.h5ad",
                "gnd_path": f"/data/Ajita/Spatial/Datasets/Spatial_Domain/Simulated_Breast_Cancer/{dataset_id}/simulated_ground_truth_{dataset_id}.csv"
            }

        elif dataset_name.startswith("simulated_liver_cancer"):
            dataset_id = dataset_name.replace("simulated_liver_cancer", "")
            return {
                "st_path": f"/data/Ajita/Spatial/Datasets/Spatial_Domain/Simulated_Liver_Cancer/simulated_gene_expression_{dataset_id}_average.h5ad",
                "gnd_path": f"/data/Ajita/Spatial/Datasets/Spatial_Domain/Simulated_Liver_Cancer/simulated_ground_truth_{dataset_id}.csv"
            }

        elif dataset_name.startswith("simulated_intestine"):
            dataset_id = dataset_name.replace("simulated_intestine", "")
            return {
                "st_path": f"/data/Ajita/Spatial/Datasets/Spatial_Domain/Simulated_Intestine/{dataset_id}/simulated_gene_expression_average.h5ad",
                "gnd_path": f"/data/Ajita/Spatial/Datasets/Spatial_Domain/Simulated_Intestine/{dataset_id}/simulated_ground_truth_{dataset_id}.csv"
            }

        elif dataset_name == "simulated_chicken_heart":
            return {
                "st_path": "/data/Ajita/Spatial/Datasets/Spatial_Domain/Simulated_ChickenHeart/simulated_gene_expression_average.h5ad",
                "gnd_path": "/data/Ajita/Spatial/Datasets/Spatial_Domain/Simulated_ChickenHeart/simulated_ground_truth.csv"
            }

        elif dataset_name == "simulated_prostate_cancer":
            return {
                "st_path": "/data/Ajita/Spatial/Datasets/Spatial_Domain/Simulated_Prostate_Cancer/simulated_gene_expression_average.h5ad",
                "gnd_path": "/data/Ajita/Spatial/Datasets/Spatial_Domain/Simulated_Prostate_Cancer/simulated_ground_truth_prostate_cancer.csv"
            }

        elif dataset_name == "simulated_cerebellum":
            return {
                "st_path": "/data/Ajita/Spatial/Datasets/Spatial_Domain/Simulated_Cerebellum/simulated_gene_expression_average.h5ad",
                "gnd_path": "/data/Ajita/Spatial/Datasets/Spatial_Domain/Simulated_Cerebellum/simulated_ground_truth_Cerebellum.csv"
            }

        else:
            raise ValueError(f"Incorrect input: {dataset_name}")

    # Build dictionary for all datasets
    return {name: _single_dataset_paths(name) for name in dataset_names}
```

```python
def build_pred_paths(dataset_names, method_names):
    """
    Build a nested dict mapping dataset_name -> method_name -> prediction file path.

    Parameters
    ----------
    dataset_names : list[str]
        List of dataset names.
    method_names : list[str]
        List of method names.

    Returns
    -------
    dict
        {dataset_name: {method_name: prediction_path}}
    """
    pred_paths = {}

    for dataset_name in dataset_names:
        pred_paths[dataset_name] = {}
        
        # Determine folder and dataset_id based on dataset_name
        if dataset_name.startswith("DLPFC"):
            dataset_id = dataset_name.replace("DLPFC", "")
            folder = "DLPFC"
        elif dataset_name.startswith("embryo"):
            dataset_id = dataset_name.replace("embryo", "")
            folder = "Embryo"
        elif dataset_name.startswith("MERFISH_brain"):
            dataset_id = dataset_name.replace("MERFISH_brain", "")
            folder = "MERFISH_brain"
        elif dataset_name == "simulated_breast_atlas":
            dataset_id = 0
            folder = "simulated_breast_atlas_ST8"
        elif dataset_name.startswith("simulated_breast_cancer"):
            dataset_id = dataset_name.replace("simulated_breast_cancer", "")
            folder = f"simulated_breast_cancer/{dataset_id}"
        elif dataset_name.startswith("simulated_intestine"):
            dataset_id = dataset_name.replace("simulated_intestine", "")
            folder = f"simulated_intestine/{dataset_id}"
        elif dataset_name.startswith("simulated_kidney_cancer"):
            dataset_id = dataset_name.replace("simulated_kidney_cancer", "")
            folder = "simulated_kidney_cancer"
        elif dataset_name.startswith("simulated_liver_cancer"):
            dataset_id = dataset_name.replace("simulated_liver_cancer", "")
            folder = "simulated_liver_cancer"
        else:
            folder = dataset_name
            dataset_id = 0

        # Now build paths for each method
        for method_name in method_names:
            if dataset_id:
                path1 = f"/data/keerthana/Spatial/spatial_domain/final_outputs/{folder}/final_domain_outputs/output_{method_name}_{dataset_id}.csv"
                path2 = f"/data/keerthana/Spatial/spatial_domain/final_outputs/{folder}/final_domain_outputs/output_{dataset_id}_{method_name}.csv"
                if os.path.exists(path1):
                    path = path1
                elif os.path.exists(path2):
                    path = path2
                else:
                    path = f"/data/keerthana/Spatial/spatial_domain/final_outputs/{folder}/final_domain_outputs/output_{method_name}.csv"
            else:
                path = f"/data/keerthana/Spatial/spatial_domain/final_outputs/{folder}/final_domain_outputs/output_{method_name}.csv"

            pred_paths[dataset_name][method_name] = path

    return pred_paths
```
