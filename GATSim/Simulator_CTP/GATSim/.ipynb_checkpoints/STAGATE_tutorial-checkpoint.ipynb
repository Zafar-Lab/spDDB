```python
section_id = '151674'
```

```python
input_dir = "/home/ajita/Documents/Spatial/Datasets/DLPFC/" + section_id + "/"
adata = sc.read_visium(path=input_dir, count_file=section_id+'_filtered_feature_bc_matrix.h5')
adata.var_names_make_unique()
```

```python
adata
```

```python
#Normalization
sc.pp.highly_variable_genes(adata, flavor="seurat_v3", n_top_genes=3000)
sc.pp.normalize_total(adata, target_sum=1e4)
sc.pp.log1p(adata)
```

```python
# read the annotation
Ann_df = pd.read_csv("/home/ajita/Documents/Spatial/Datasets/DLPFC/151674/metadata.tsv", delimiter = "\t")["layer_guess_reordered"]

print (Ann_df)
```

```python
adata.obs['Ground Truth'] = Ann_df.values
```

```python
plt.rcParams["figure.figsize"] = (3, 3)
sc.pl.spatial(adata, img_key="hires", color=["Ground Truth"])
```

```python
STAGATE.Cal_Spatial_Net(adata, rad_cutoff=150)
STAGATE.Stats_Spatial_Net(adata)
```

```python
tf.compat.v1.disable_eager_execution()
adata = STAGATE.train_STAGATE(adata, alpha=0, save_reconstrction=True)
```

```python

sc.pp.neighbors(adata, use_rep='STAGATE')
sc.tl.umap(adata)
adata = STAGATE.mclust_R(adata, used_obsm='STAGATE', num_cluster=7)
```

```python
obs_df = adata.obs.dropna()
ARI = adjusted_rand_score(obs_df['mclust'], obs_df['Ground Truth'])
print('Adjusted rand index = %.2f' %ARI)
```

```python
print (adata.X.shape)
print (adata.obsm["STAGATE"].shape)
print (adata)
```

```python
plt.rcParams["figure.figsize"] = (3, 3)
sc.pl.umap(adata, color=["mclust", "Ground Truth"], title=['STAGATE (ARI=%.2f)'%ARI, "Ground Truth"])
```
